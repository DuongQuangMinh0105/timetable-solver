```python
from ortools.sat.python import cp_model

classes = ["L1", "L2", "L3"]
rooms = ["R1", "R2"]
slots = [1, 2, 3, 4]

model = cp_model.CpModel()
x = {(c, r, t): model.NewBoolVar(f"x_{c}_{r}_{t}")
     for c in classes for r in rooms for t in slots}

# Mỗi lớp được xếp đúng 1 (phòng, tiết)
for c in classes:
    model.AddExactlyOne(x[c, r, t] for r in rooms for t in slots)

# Mỗi (phòng, tiết) tối đa 1 lớp
for r in rooms:
    for t in slots:
        model.AddAtMostOne(x[c, r, t] for c in classes)

solver = cp_model.CpSolver()
status = solver.Solve(model)
print(solver.StatusName(status))

if status in (cp_model.OPTIMAL, cp_model.FEASIBLE):
    for c in classes:
        for r in rooms:
            for t in slots:
                if solver.Value(x[c, r, t]):
                    print(f"{c} -> {r}, tiết {t}")
```
